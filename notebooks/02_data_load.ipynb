```python
import pandas as pd
import numpy as np

X = pd.read_csv("../data/processed/secom_features_raw.csv")
```

```python
# Calculate missing ratio per sensor
missing_ratio = X.isna().mean()

# Drop sensors with > 80% missing values
threshold = 0.80
X_reduced = X.loc[:, missing_ratio <= threshold]

print(f"Original features: {X.shape[1]}")
print(f"Remaining features: {X_reduced.shape[1]}")
```

```python
X_imputed = X_reduced.fillna(X_reduced.median())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = pd.DataFrame(
    scaler.fit_transform(X_imputed),
    columns=X_imputed.columns
)
```

```python
X_scaled.to_csv("../data/processed/secom_features_clean.csv", index=False)

print("Saved cleaned and scaled features.")
```
